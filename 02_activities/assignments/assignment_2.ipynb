#### Starter Code for Question 1

```python
from collections import deque
```

```python
# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right

def is_duplicate(root: TreeNode) -> int:
  # TODO
  if not root:
      return 
    
  # implement BFS
  queue = deque([root])
  value_tracker = {}
  
  while queue:
      current_node = queue.popleft()

      if current_node.val in value_tracker:
        return current_node.val
      
      value_tracker[current_node.val] = True

      if current_node.left:
        queue.append(current_node.left)
            
      if current_node.right:
          queue.append(current_node.right)
            
  return -1
```

```python
# Your answer here
New example:
Input: {([])[()]}
Output: True

Walkthrough of an example that my partner made is:
Input: s = "}([{}])"
Output: False

The string starts with a closing curly bracket. 
Since it has no closing corresponding bracket that came before then the string is not valid.
Walking through my partner's code, the can_pop method will return False since the stack will be empty.

```


-   Copy the solution your partner wrote. 


```python
# Your answer here
class BracketStack:
    brackets_dict = {
        '(': ')',
        '[': ']',
        '{': '}'
    }

    def __init__(self):
        self.stack = []

    def is_open_bracket(self, c: str) -> bool:
        return c in self.brackets_dict.keys()
    
    def is_close_bracket(self, c: str) -> bool:
        return c in self.brackets_dict.values()
    
    def is_stack_empty(self):
        return len(self.stack) == 0
    
    def can_pop(self, closing_bracket: str) -> bool:
        if not self.is_stack_empty():
            last_in_stack = self.stack[-1]
            if self.is_open_bracket(last_in_stack) and self.brackets_dict[last_in_stack] == closing_bracket:
                return True

        return False

    def push(self, open_bracket: str):
        self.stack.append(open_bracket)

    def pop(self, closing_bracket: str):
        self.stack.pop()
        
    def is_valid_brackets(self, s: str) -> bool:
        for c in s:
            if self.is_open_bracket(c):
                self.push(c)
            elif self.is_close_bracket(c) and self.can_pop(c):
                self.pop(c)
            else:
                return False
            
        # Check if the stack is empty at the end
        if not self.is_stack_empty():
            return False
        return True
    
# Original block of code
def is_valid_brackets(s: str) -> bool:
    stack = BracketStack()
    return stack.is_valid_brackets(s)

example_list = [
    "([]{})",
    "([)]",
    "()[]{}",
    "[{]}",
    "(((",
    "{([])[()]}", # new example
]
for example in example_list:
    print(f"\"{example}\" : {is_valid_brackets(example)}")
```


-   Explain why their solution works in your own words.


```python
# Your answer here
They used an elegant solution by using a stack. 
Opening bracket is pushed onto the stack 
Closing bracket triggers a pop, but first there is a check for whether the closing bracket has a corresponding opening bracket on top of the stack
Finally there is a check if the stack is empty, meaning, all opening and closing brackets were checked and no other element remains on the stack
```


-   Explain the problem’s time and space complexity in your own words.


```python
# Your answer here
There is a loop of the list, so the time complexity is O(n)
The space complexity is O(n) because if all elements are opened brackets, all of the elements will be stacked on top of each other - worst case scenario.
For cases like '()[]{}', the space complexity could be something like O(1) because as soon as an element is stacked, a closing bracket trigger it to be popped off of the stack - best case scenario.
```
